# response check
Checks ResponseDB loading, one response matrix, and a contraction.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.nuclear import response as mod

assert "core.engine" not in inspect.getsource(mod)
db = mod.ResponseDB(str(ROOT / "data" / "responses" / "responses_all.json"))
resp = db.get("Cs133")
W = resp.W_np("M", 0.01)
assert np.asarray(W).shape == (2, 2)
assert np.asarray(W).dtype == np.dtype("float64")
val = resp.contract_np("M", 0.01, an=1.0, ap=0.5)
assert np.isfinite(val)
print({"module": "response", "checks": 5, "status": "PASS"})
```
