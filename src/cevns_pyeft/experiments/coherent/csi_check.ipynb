# csi check
Checks CsI bins and representative efficiency values.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.experiments.coherent import csi as mod

assert "core.engine" not in inspect.getsource(mod)
pe = np.array([5.0, 12.0, 20.0], dtype=float)
eff = np.asarray(mod.efficiency(pe), dtype=float)
assert eff.shape == pe.shape and eff.dtype == np.dtype("float64")
assert np.isfinite(eff).all() and np.all(eff >= 0.0)
assert mod.CSI2017_BINS_25.ndim == 1 and mod.CSI2017_WIDTHS_25.size == 25
print({"module": "csi", "checks": 5, "status": "PASS"})
```
