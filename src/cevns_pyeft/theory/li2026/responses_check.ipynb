# responses check
Checks Li2026 particle-side R matrices for fixed ci and kinematics.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.theory.li2026 import responses as mod

assert "core.engine" not in inspect.getsource(mod)
ci_basis = {i: (0.01 + 0j, -0.02 + 0j) for i in [1,2,3,4,5,6,8,9,10,11,13,14]}
kin = mod.RKinematics(Enu=0.03, qabs=0.02, m_A=120.0, m_N=0.939)
scalar = mod.responses_R(ci_basis, kin, tau=0, taup=1)
mats = mod.compute_R_matrices(ci_basis, E_nu=0.03, qabs=0.02, mA=120.0, mN=0.939)
assert set(scalar) == {"M", "SigmaPP", "SigmaP", "PhiPP", "PhiPT", "Delta", "MPhiPP", "SigmaPDelta"}
assert all(np.asarray(v).shape == (2, 2) for v in mats.values())
assert np.isfinite(np.array([complex(v).real for v in scalar.values()], dtype=float)).all()
print({"module": "responses", "checks": 5, "status": "PASS"})
```
