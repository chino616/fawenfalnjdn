# cross_section check
Checks one Li2026 d sigma / d ER calculation at fixed kinematics.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.nuclear.response import ResponseDB
from cevns_pyeft.theory.li2026 import cross_section as mod

assert "core.engine" not in inspect.getsource(mod)
resp = ResponseDB(str(ROOT / "data" / "responses" / "responses_all.json")).get("Cs133")
left_hat = {(6, 1, "u"): 1.0e-5 + 0j, (6, 1, "d"): -2.0e-5 + 0j}
ds = mod.dsigma_dER_li2026_cm2_per_GeV(resp, E_nu_GeV=0.03, ER_keV=10.0, left_hat=left_hat)
assert isinstance(ds, float)
assert np.isfinite(ds)
assert ds >= 0.0
print({"module": "cross_section", "checks": 4, "status": "PASS"})
```
