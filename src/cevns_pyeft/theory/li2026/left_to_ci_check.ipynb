# left_to_ci check
Checks fixed LEFT coefficients mapped to proton, neutron, and isospin ci dictionaries.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
import importlib
mod = importlib.import_module("cevns_pyeft.theory.li2026.left_to_ci")

assert "core.engine" not in inspect.getsource(mod)
left_hat = {(6, 1, "u"): 1.0e-5 + 0j, (6, 1, "d"): -2.0e-5 + 0j, (7, 4, "u"): 3.0e-7 + 0j}
ci = mod.left_to_ci(left_hat, E_nu=0.03, qabs=0.02, mA=120.0)
assert sorted(ci.p) == list(range(1, 15))
assert sorted(ci.n) == list(range(1, 15))
assert sorted(ci.iso) == list(range(1, 15))
assert np.isfinite(np.array([complex(ci.p[1]).real, complex(ci.n[1]).real, complex(ci.iso[1][0]).real], dtype=float)).all()
print({"module": "left_to_ci", "checks": 5, "status": "PASS"})
```
