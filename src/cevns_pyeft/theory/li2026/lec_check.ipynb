# lec check
Checks Li2026 default LEC dataclass and dictionary conversion.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.theory.li2026 import lec as mod

assert "core.engine" not in inspect.getsource(mod)
d = mod.as_dict(mod.DEFAULT_LECS)
assert isinstance(mod.DEFAULT_LECS, mod.Li2026LECs)
assert {"m_u", "m_d", "g_A", "m_N"}.issubset(d)
assert np.isfinite(np.array(list(d.values()), dtype=float)).all()
print({"module": "lec", "checks": 4, "status": "PASS"})
```
