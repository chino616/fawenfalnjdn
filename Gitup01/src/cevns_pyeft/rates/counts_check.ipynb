# counts check
Checks one minimal binned expected-count calculation.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.nuclear.response import ResponseDB
from cevns_pyeft.experiments.coherent.sns import loadNeutrinoEmpiricalFluxSNS
from cevns_pyeft.experiments.coherent.slow_counts_csi2021 import efficiency
from cevns_pyeft.rates import counts as mod

assert "core.engine" not in inspect.getsource(mod)
resp = ResponseDB(str(ROOT / "data" / "responses" / "responses_all.json")).get("Cs133")
flux = loadNeutrinoEmpiricalFluxSNS("CsI2021", num_points=20)
bins = np.array([0.001, 8.001, 12.001], dtype=float)
out = mod.bin_events_csi(resp, flux, bins, efficiency=efficiency, include_sm=True, n_ER=5, n_Enu=20, n_PE_bin=4)
assert out.shape == (2,) and out.dtype == np.dtype("float64")
assert np.isfinite(out).all() and np.all(out >= 0.0)
print({"module": "counts", "checks": 5, "status": "PASS"})
```
