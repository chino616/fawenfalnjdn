# differential check
Checks differential cross section on a fixed recoil and minimal neutrino-energy array.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.nuclear.response import ResponseDB
from cevns_pyeft.rates import differential as mod

assert "core.engine" not in inspect.getsource(mod)
resp = ResponseDB(str(ROOT / "data" / "responses" / "responses_all.json")).get("Cs133")
enu = np.array([10.0, 30.0], dtype=float)
res = mod.differential_cross_section_nn(resp, 10.0, enu, include_sm=True)
arr = np.asarray(res.dsigma_cm2_per_keV, dtype=float)
assert arr.shape == enu.shape and arr.dtype == np.dtype("float64")
assert np.isfinite(arr).all() and np.all(arr >= 0.0)
print({"module": "differential", "checks": 5, "status": "PASS"})
```
