```python
import inspect, json, sys
from pathlib import Path
import numpy as np
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src').exists())
sys.path.insert(0, str(root / 'src'))
from cevns_pyeft.kernels import reconstruction as mod
assert 'core.engine' not in inspect.getsource(mod)
ref = root / 'tests/reference_fast_split'
full = np.load(ref / 'csi2021_full_kernel.npy')
coeff = np.zeros(full.shape[3], dtype=complex); coeff[0]=1; coeff[2]=0.3+0.2j
raw = mod.reconstruct_real_events_from_full_kernel(full, coeff)
cmp = mod.reconstruct_real_events_from_pair_templates(diag=np.load(ref/'csi2021_diag.npy'), pair_re=np.load(ref/'csi2021_pair_re.npy'), pair_im=np.load(ref/'csi2021_pair_im.npy'), pair_index=np.load(ref/'csi2021_pair_index.npy'), coeffs=coeff)
np.testing.assert_allclose(raw, cmp, rtol=1e-14, atol=1e-8)
print({'module': 'reconstruction', 'shape': raw.shape, 'status': 'PASS'})
```
