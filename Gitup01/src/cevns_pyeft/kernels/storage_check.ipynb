```python
import inspect, json, sys
from pathlib import Path
import numpy as np
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src').exists())
sys.path.insert(0, str(root / 'src'))
import tempfile
from cevns_pyeft.experiments.coherent.fast_kernel_csi2021 import build_fast_kernel_csi2021
from cevns_pyeft.kernels import storage as mod
assert 'core.engine' not in inspect.getsource(mod)
k = build_fast_kernel_csi2021(root/'data/responses/responses_all.json', n_er=3, n_enu=6, n_pe_sub=3)
with tempfile.TemporaryDirectory() as td:
    p = mod.save_fastscan_template(Path(td)/'kernel.npz', k)
    loaded = mod.load_fastscan_template(p)
np.testing.assert_array_equal(loaded['full_kernel'], k['full_kernel'])
print({'module': 'storage', 'dataset': loaded['metadata']['dataset'], 'status': 'PASS'})
```
