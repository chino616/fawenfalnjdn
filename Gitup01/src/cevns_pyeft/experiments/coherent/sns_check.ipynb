# sns check
Checks prompt and delayed SNS flux on a minimal energy array.


```python
from pathlib import Path
import sys
import inspect
import numpy as np

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
assert (ROOT / "src").exists()
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
```

```python
from cevns_pyeft.experiments.coherent import sns as mod

assert "core.engine" not in inspect.getsource(mod)
flux = mod.loadNeutrinoEmpiricalFluxSNS("CsI2021", num_points=20)
E = np.array([10.0, 30.0], dtype=float)
vals = np.asarray(flux["measured"]["totalContinuous"](E), dtype=float)
mu_E0, mu_w = flux["measured"]["muLine"]
assert vals.shape == E.shape and vals.dtype == np.dtype("float64")
assert np.isfinite(vals).all() and mu_E0 > 0.0 and mu_w > 0.0
print({"module": "sns", "checks": 5, "status": "PASS"})
```
