```python
import json, sys
from pathlib import Path
import numpy as np

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src').exists())
sys.path.insert(0, str(root / 'src'))
from cevns_pyeft.experiments.coherent.slow_counts_csi2021 import PE_BINS_9, efficiency_csi2021, slow_counts_csi2021

ref = json.loads((root / 'backup' / 'pre_split_reference_slow_counts.json').read_text())['CsI2021']
result = slow_counts_csi2021(root / 'data/responses/responses_all.json', include_sm=True, n_er=5, n_enu=20, n_pe_bin=4)
print('bins', PE_BINS_9.size - 1, 'eff@12', float(efficiency_csi2021(12.0)), 'sum', float(result['counts_per_bin']['total'].sum()))
for section in ('counts_per_bin', 'counts_per_npe'):
    for key, arr in result[section].items():
        got = np.asarray(arr)
        exp = np.asarray(ref[section][key])
        assert got.dtype == np.dtype(ref['dtypes'][key])
        assert got.shape == tuple(ref['shapes'][key])
        np.testing.assert_array_equal(got, exp)
```
